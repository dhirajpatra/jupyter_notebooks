By reading a data frame where all employees and branch name and sales are there, I need to read that and find out that across that all branches the sales which, um, employee has made highest

```python
import pandas as pd

# Example DataFrame
df = pd.DataFrame({
    'employee': ['A', 'B', 'A', 'C', 'B', 'C'],
    'branch':   ['X', 'X', 'Y', 'Y', 'Y', 'X'],
    'sales':    [100, 200, 150, 300, 50, 250]
})

# Sum sales by branch and employee
branch_emp_sales = df.groupby(['branch', 'employee'], as_index=False)['sales'].sum()

# Get the employee with highest sales in each branch
top_employee_per_branch = branch_emp_sales.loc[
    branch_emp_sales.groupby('branch')['sales'].idxmax()
]

print(top_employee_per_branch)
```

```python
top_names = top_employee_per_branch[['branch', 'employee']]
print(top_names)
```

```python
import pandas as pd

# Suppose your DataFrame is called df with columns: employee, branch, sales

# 1. Sum sales by employee across all branches
emp_sales = df.groupby('employee', as_index=False)['sales'].sum()

# 2. Get the row with the highest total sales
top_employee = emp_sales.loc[emp_sales['sales'].idxmax()]

# 3. Print employee name and total sales
print("Employee:", top_employee['employee'])
print("Total Sales:", top_employee['sales'])
```

```python
top_employee_df = emp_sales.loc[[emp_sales['sales'].idxmax()]]
print(top_employee_df[['employee', 'sales']])
```
