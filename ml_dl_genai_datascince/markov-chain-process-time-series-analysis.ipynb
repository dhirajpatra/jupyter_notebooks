```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define states and mapping
states = ["Sunny", "Cloudy", "Rainy"]
state_map = {0: "Sunny", 1: "Cloudy", 2: "Rainy"}
```

```python
# Transition matrix P (Rows: Current State -> Columns: Next State)
# Order: [Sunny, Cloudy, Rainy]
P = np.array([
    [0.8, 0.1, 0.1],  # From Sunny
    [0.2, 0.6, 0.2],  # From Cloudy
    [0.3, 0.3, 0.4]   # From Rainy
])
```

```python
# Simulation Parameters
num_days = 30
initial_state = 0  # Start at 'Sunny'
```

```python
# Store simulated path
time_series = [initial_state]
current_state = initial_state
```

```python
# Run Markov Chain Simulation
np.random.seed(42)  # For reproducible output
for _ in range(num_days - 1):
    # Sample next state based on current row's transition probabilities
    next_state = np.random.choice([0, 1, 2], p=P[current_state])
    time_series.append(next_state)
    current_state = next_state
```

```python
# Print sequence as text labels
sequence_labels = [state_map[s] for s in time_series]
print("First 10 Days Sequence:")
print(" -> ".join(sequence_labels[:10]))
```

```python
# Plot the Time Series
plt.figure(figsize=(10, 4))
plt.step(range(num_days), time_series, where='post', color='navy', linewidth=2)
plt.yticks([0, 1, 2], states)
plt.xlabel("Day (Time Step t)")
plt.ylabel("Weather State")
plt.title(f"{num_days}-Day Markov Chain Weather Simulation")
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
