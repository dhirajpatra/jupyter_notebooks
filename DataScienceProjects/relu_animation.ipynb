```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.ndimage import convolve

# 1. Create a synthetic image with a clear structure (a circle)
x, y = np.ogrid[-100:100, -100:100]
mask = x**2 + y**2 < 60**2
image = np.zeros((200, 200))
image[mask] = 1.0

# 2. Apply a Sobel filter to create positive and negative feature maps
sobel_y = np.array([[-1, -2, -1], 
                    [ 0,  0,  0], 
                    [ 1,  2,  1]])
feature_map = convolve(image, sobel_y)

# 3. Setup the plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
fig.suptitle("Applying ReLU Activation Function to Feature Map", fontsize=14, fontweight='bold')

# Left side: Original Feature Map (with negative values)
im1 = ax1.imshow(feature_map, cmap='coolwarm')
ax1.set_title("Input Feature Map\n(Contains ± values)")
fig.colorbar(im1, ax=ax1, fraction=0.046, pad=0.04)

# Right side: Output being transformed by ReLU
# Start with a copy of the input map to animate the transition row by row
animated_map = feature_map.copy()
im2 = ax2.imshow(animated_map, cmap='coolwarm', vmin=feature_map.min(), vmax=feature_map.max())
ax2.set_title("ReLU Output\nmax(0, x)")
fig.colorbar(im2, ax=ax2, fraction=0.046, pad=0.04)

# Add a visual scanning line tracker
scan_line = ax1.axhline(0, color='yellow', lw=2)

# 4. Animation update function
def update(row):
    # For the current row, apply ReLU: change all negative numbers to 0
    animated_map[row, :] = np.maximum(0, feature_map[row, :])
    
    # Update visual data
    im2.set_array(animated_map)
    scan_line.set_ydata([row, row])
    
    return im2, scan_line

# 5. Run and save the animation
# frames matches the height of the image (200 pixels)
ani = FuncAnimation(fig, update, frames=feature_map.shape[0], interval=20, blit=True)

# Save as a GIF
ani.save('relu_activation.gif', writer='pillow', fps=30)
plt.close()
print("Animation saved successfully as 'relu_activation.gif'!")
```
